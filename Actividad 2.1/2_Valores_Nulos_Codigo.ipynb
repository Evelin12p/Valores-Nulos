**Métodos de Sustitución de Valores Nulos**

```python
data_limpia = data.copy()
```

```python
# Quitar espacios en blanco al inicio y al final de los nombres de las columnas
data_limpia.columns = data_limpia.columns.str.strip()
```

```python
print(data_limpia.dtypes)
```

```python
data_limpia['Potencial de compra'] = pd.to_numeric(data_limpia['Potencial de compra'], errors='coerce')
```

```python
print(data_limpia.dtypes)
```

```python
# 1. Rellenar los valores nulos de la columna 'BDC' 

data_limpia['BDC'] = data_limpia['BDC'].fillna("Sin BDC")
```

Elegoi la opción de sustitución por un valor de texto específico "Sin BDC", debido a que es una variable categórica y no es posible aplicar medidas estadísticas como la media o mediana. Se evita utilizar ffill o bfill porque el valor de una fila no necesariamente corresponde al de la fila siguiente o anterior.

```python
data_limpia.isnull().sum()
```

```python
# 2 Rellenar los valores nulos de la columna Fecha de Cita
data_limpia['Fecha'] = data_limpia['Fecha'].ffill()
```

```python
print(data_limpia['Fecha'])
```

Se uso para preserva la secuencia cronológica lógica del registro de fechas, garantizando que cada lead quede agrupado en el bloque temporal más cercano sin alterar la continuidad del conjunto de datos.

```python
data_limpia.isna().sum()
```

```python
# 3 Rellenar valores nulos de la columna Nombre de Cliente 
data_limpia = data_limpia.dropna(subset=['Nombre Cliente'])
```

Decidi borrar las filas porque no representan interacciones reales ni citas, entonces son errores de formato/diseño del reporte visual original. Al quitarlos evitamos confuciones e evita Sesgos en Métricas Si se conservan o se imputan y asi no arruinan los calculos de tasa de conversión de la agencia.

```python
# 4 limpiar valores Nulos de la columna Estatus de Lead  
data_limpia['Estatus de Lead'] = data_limpia['Estatus de Lead'].fillna("Sin Estatus")
```

Decidi rellenarlo con "Sin Estatus" ya que no se puede determinar el estatus de un lead sin información.  

```python
data_limpia.isnull().sum()
```

```python
# 5 limpiar valores Nulos de la columna Potencial de compra con mediana 

data_limpia['Potencial de compra'] = data_limpia['Potencial de compra'].fillna(data_limpia['Potencial de compra'].median())
```

se relleno con mediana ya que es una columna numerica y por lo mismo usar mediana es menos sensible a valores extremos que la media. De esta manera se pueden conservar los registros sin modificar excesivamente la distribución de la variable. 

```python
data_limpia.isnull().sum()
```

```python
# 6 Valores nulos de las columnas PDM, SDC, Venta y Asesor Asignado 
data_limpia['PDM'] = data_limpia['PDM'].fillna(0.0)
data_limpia['SDC'] = data_limpia['SDC'].fillna(0.0)
data_limpia['Venta'] = data_limpia['Venta'].fillna(0.0)
data_limpia['Asesor Asignado'] = data_limpia['Asesor Asignado'].fillna("Asesor no identificado")
```

Se utilizó la sustitución por el valor numérico 0, debido a que PDM, SDC y Venta son variables binaria cuyos valores existentes son 0 y 1. El valor 0 permite conservar la estructura de las variables y evitar introducir valores que no pertenecen a sus categorías originales.
Para Asesor asignado utilizó un valor de texto específico "Sin Asesor" para representar los registros donde no existe un asesor asignado. No se utilizó ffill ni bfill, ya que el asesor corresponde a cada cliente y no necesariamente es el mismo que el de los registros contiguos.

```python
data_limpia.isnull().sum()
```

```python
# quitar nulos de las columnas de preguntas
columnas_preguntas = ['¿Por que no ha visitado la agencia?', '¿Por que no hay solicitud de credito?', '¿Por que no hubo prueba de manejo ?','¿Por que no hubo proceso wow?', '¿Por que se asigno antes de vistar piso?']

data_limpia[columnas_preguntas] = data_limpia[columnas_preguntas].fillna("Sin información")
```

Para las variables correspondientes a preguntas sobre el comportamiento o situación del cliente, se utilizó la sustitución de valores nulos por "Sin información". Estas variables son de tipo texto y sus respuestas representan situaciones particulares de cada cliente, por lo que no es adecuado utilizar métodos estadísticos como la media o mediana, ni métodos como ffill o bfill, ya que podrían asignar una respuesta de otro registro que no corresponde, al utilizar "Sin información" se conservan los registros y ya se diferencia los valores que no cuentan con una respuesta registrada

```python
data_limpia.isnull().sum()
```

```python
data_limpia.to_csv('Citas_Digital_Limpio.csv', index=False)
```
