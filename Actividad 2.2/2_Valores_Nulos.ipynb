**VALORES NULOS**

```python
# Archivo Base de Tlaxcala
df = pd.read_excel('Base_de_Tlaxcala.xlsx', index_col=False)
df.head(5)
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df_limpio = df.copy()
```

```python
# .str.strip() elimina espacios al inicio y final
# .str.upper() convierte todo a MAYÚSCULAS para homogenizar

df_limpio['Asesor'] = df_limpio['Asesor'].astype(str).str.strip().str.upper()
```

Identifique que habia Nombres escritos con minusculización y otros con mayúsculas. La conversión a mayúsculas es solo para uniforma el catálogo de personal y se vea mejor.

```python
df_limpio["Producto"]= df_limpio["Producto"].fillna('Sin información') 
```

```python
df_limpio["Producto"].value_counts()
```

Al existir un único registro faltante en Producto correspondiente a un lead con estatus Return (devuelto/inconcluso), la imputación asignaría una preferencia de vehículo no confirmada. Se etiqueta como 'Sin información' para mantener el conteo total de leads

```python
df_limpio.isnull().sum()
```

```python
df_limpio = df_limpio.drop(columns=['Telefono', 'Correo'])
```

Se elimina porque presentan un 100% de valores nulos. Mantener variables totalmente vacías solo consume espacio en memoria y genera ruido en los modelos analíticos.
