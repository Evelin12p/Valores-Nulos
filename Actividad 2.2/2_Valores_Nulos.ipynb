```python
df_sdc_limpio = df_sdc.copy()
```

```python
print(df_sdc_limpio.shape)
```

```python
meses_es = ["Enero","Febrero","Marzo","Abril","Mayo","Junio","Julio",
            "Agosto","Septiembre","Octubre","Noviembre","Diciembre"]

encabezados_correctos = []
mes_num = 0
anio = 24
for i in range(31):
    encabezados_correctos.append(f"{meses_es[mes_num]}, {anio}")
    mes_num += 1
    if mes_num == 12:
        mes_num = 0
        anio += 1

print(encabezados_correctos)
```

```python
df_sdc_limpio.iloc[2, 1:32] = encabezados_correctos
df_sdc_limpio.iloc[32, 1:32] = encabezados_correctos
df_sdc_limpio.iloc[40, 1:32] = encabezados_correctos
```

```python
print(df_sdc_limpio.iloc[2, 1:32].tolist())
print(df_sdc_limpio.iloc[32, 1:32].tolist())
print(df_sdc_limpio.iloc[40, 1:32].tolist())
```

tenían el mismo error vienen copiadas del mismo encabezado original con "Diciembre, 25" duplicado. sobrescrib la fila completa de una vez con la secuencia correcta y verificada para así garantizamos consistencia total.

```python
df_sdc_limpio.iloc[16:30, 1:32] = df_sdc_limpio.iloc[16:30, 1:32].replace("-", 0)
df_sdc_limpio.iloc[16:30, 1:32] = df_sdc_limpio.iloc[16:30, 1:32].fillna(0)
```

```python
print(df_sdc_limpio.iloc[16:30, 1:32])
```

ambas representan lo mismo (cero solicitudes ese mes para esa financiera). Al unificar todo a 0 (número), la columna completa queda numérica de verdad 

```python
import numpy as np

df_sdc_limpio.iloc[33:37, 1:17] = np.nan  
df_sdc_limpio.iloc[41:45, 1:17] = np.nan   
```

aunque ya venían como NaN en el archivo original, hacerlo explícito por código deja documentado en el propio script por qué esas celdas son nulas (categoría no medida aún) y protege contra que, si el archivo fuente cambia o alguien vuelve a cargarlo, esas celdas queden accidentalmente en 0 o '-' en vez de NaN.

```python
df_sdc_limpio.iloc[16:30, 0] = df_sdc_limpio.iloc[16:30, 0].str.strip().str.upper()
print(df_sdc_limpio.iloc[16:30, 0].tolist())
```

 La conversión a mayúsculas es solo para uniforma el catálogo de personal y se vea mejor.

```python
total_general = df_sdc_limpio.iloc[16:30, 32].sum()
df_sdc_limpio.iloc[16:30, 33] = df_sdc_limpio.iloc[16:30, 32] / total_general
df_sdc_limpio.iloc[15, 33] = None
print(df_sdc_limpio.iloc[16:30, 33].sum())
```

```python
df_sdc_limpio.to_csv("SDC_limpio.csv", index=False, header=False)
```

```python
bloque1_tidy = df_sdc_limpio.iloc[3:8, 0:32].copy()
bloque1_tidy.columns = ["Indicador"] + encabezados_correctos
print(bloque1_tidy)
```

```python
#Fomato tidy 
bloque1_tidy = bloque1_tidy.melt(id_vars="Indicador", var_name="Mes", value_name="Valor")
print(bloque1_tidy)
```

```python
bloque2_tidy = df_sdc_limpio.iloc[16:30, 0:32].copy()
bloque2_tidy.columns = ["Financiera"] + encabezados_correctos
print(bloque2_tidy)
```

```python
bloque2_tidy = bloque2_tidy.melt(id_vars="Financiera", var_name="Mes", value_name="Valor")
print(bloque2_tidy)
```

```python
bloque3_tidy = df_sdc_limpio.iloc[33:37, 0:32].copy()
bloque3_tidy.columns = ["Motivo"] + encabezados_correctos
print(bloque3_tidy)
```

```python
bloque3_tidy = bloque3_tidy.melt(id_vars="Motivo", var_name="Mes", value_name="Valor")
print(bloque3_tidy)
```

```python
bloque4_tidy = df_sdc_limpio.iloc[41:45, 0:32].copy()
bloque4_tidy.columns = ["Motivo"] + encabezados_correctos
print(bloque4_tidy)
```

```python
bloque4_tidy = bloque4_tidy.melt(id_vars="Motivo", var_name="Mes", value_name="Valor")
print(bloque4_tidy)
```

```python
with pd.ExcelWriter("SDC_tidy.xlsx") as writer:
    bloque1_tidy.to_excel(writer, sheet_name="Embudo", index=False)
    bloque2_tidy.to_excel(writer, sheet_name="Financieras", index=False)
    bloque3_tidy.to_excel(writer, sheet_name="Motivos_Rechazo", index=False)
    bloque4_tidy.to_excel(writer, sheet_name="Motivos_No_Formalizacion", index=False)
```

Se paso a un formato tydy porque el formato ancho (meses como columnas) obliga a reescribir la misma lógica de agrupar/sumar cada vez que se cambia de pregunta En tidy una fila por dato individual lo que facilita cualquier groupby, gráfica o filtro funciona directo, sin transformar la tabla de nuevo cada vez.
